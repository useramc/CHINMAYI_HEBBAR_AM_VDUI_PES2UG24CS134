## Task 5 — Compare PCA, t-SNE, and UMAP (Complaint #4)

**Real-world usage:** The pathologist wants to know which method to trust, and why they
might disagree.

**Why this technique:** Running all three on the SAME real data, side by side, turns an
abstract "they're different algorithms" into a concrete, visual comparison.

**TODO:**
1. Standardize `X_raw` using `StandardScaler`.
2. Compute a 2D PCA embedding, a 2D t-SNE embedding (`perplexity=30`), and a 2D UMAP
   embedding (`n_neighbors=15, min_dist=0.1`) — all with `random_state=0`.
3. Plot all three side by side, colored by `diagnosis`.

```python
def pca_from_scratch(X, k):

    # TODO 1: center AND standardize X
    X_centered = X - np.mean(X, axis=0)
    X_scaled = X_centered / np.std(X, axis=0, ddof=1)

    # TODO 2: compute the covariance matrix of X_scaled
    cov = np.cov(X_scaled, rowvar=False)

    # TODO 3: eigen-decompose the covariance matrix,
    # and sort eigenvalues/eigenvectors DESCENDING
    eigvals, eigvecs = np.linalg.eigh(cov)

    order = np.argsort(eigvals)[::-1]

    eigvals = eigvals[order]
    eigvecs = eigvecs[:, order]

    # TODO 4: keep only the top k eigenvectors
    # and project X_scaled onto them
    top_eigvecs = eigvecs[:, :k]
    Z = X_scaled @ top_eigvecs

    return Z, eigvals, eigvecs


Z, eigvals, eigvecs = pca_from_scratch(X_raw, k=2)

fig, ax = plt.subplots(figsize=(6, 5))

for label, color in [("B", "#3b6ea5"), ("M", "#b5432e")]:
    mask = y == label
    ax.scatter(
        Z[mask, 0],
        Z[mask, 1],
        s=15,
        alpha=0.6,
        color=color,
        label=label
    )

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_title("Your from-scratch PCA")
ax.legend()

plt.show()
```

**Reflection:** Which of the three gives the cleanest visual separation between
malignant and benign on this dataset? Per the lecture's comparison table, is that method

(interpretability, speed, local vs. global structure)?

_Your answer:_
UMAP (and t-SNE) gives the cleanest visual separation, pulling malignant and benign cases into tightly separated, non-linear clusters, whereas PCA leaves noticeable overlap along its boundaries.

However, no single method is universally "best." UMAP and t-SNE excel at uncovering complex local relationships and visual clustering, but they are non-deterministic, computationally expensive, and distort global distances. In contrast, PCA is fast, deterministic, and preserves global data structure while maintaining direct feature interpretability through loadings, making it preferable when global context or exact variable contributions matter more than raw cluster separation.
